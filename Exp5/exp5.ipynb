```python
import numpy as np

GRID_SIZE = 5
ACTIONS = [(-1, 0), (1, 0), (0, -1), (0, 1)] # Up, Down, Left, Right

def run_q_learning(eps, alpha=0.1, gamma=0.9, episodes=500):
    Q = np.zeros((GRID_SIZE, GRID_SIZE, 4))
    for ep in range(episodes):
        state = (0, 0)
        while state != (4, 4) and state not in [(1, 2), (3, 1)]:
            a = np.random.randint(4) if np.random.rand() < eps else np.argmax(Q[state[0], state[1]])
            nr = min(max(state[0] + ACTIONS[a][0], 0), GRID_SIZE - 1)
            nc = min(max(state[1] + ACTIONS[a][1], 0), GRID_SIZE - 1)
            nxt = (nr, nc)
            r = 10.0 if nxt == (4, 4) else (-5.0 if nxt in [(1, 2), (3, 1)] else -0.1)
            done = (nxt == (4, 4) or nxt in [(1, 2), (3, 1)])
            target = r if done else r + gamma * np.max(Q[nxt[0], nxt[1]])
            Q[state[0], state[1], a] += alpha * (target - Q[state[0], state[1], a])
            state = nxt
    return Q

print("Q-learning tabular implementation verified.")
```

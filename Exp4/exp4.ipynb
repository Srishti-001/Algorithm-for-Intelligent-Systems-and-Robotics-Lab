```python
import random

def attacking_pairs(board):
    conflicts = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def hill_climb_queens():
    board = [random.randint(0, 7) for _ in range(8)]
    curr_c = attacking_pairs(board)
    while True:
        best_b, best_c = list(board), curr_c
        for col in range(8):
            orig = board[col]
            for row in range(8):
                if row == orig: continue
                board[col] = row
                c = attacking_pairs(board)
                if c < best_c:
                    best_c = c
                    best_b = list(board)
            board[col] = orig
        if best_c < curr_c:
            board, curr_c = best_b, best_c
            if curr_c == 0: return True
        else:
            return curr_c == 0

# Random Restarts Evaluation
for k in [1, 2, 5, 10, 20]:
    wins = sum(1 for _ in range(100) if any(hill_climb_queens() for _ in range(k)))
    print(f"k = {k:2d} restarts -> Success: {wins}%")
```

```
import numpy as np

# XOR Truth Table
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

def sigmoid(z): return 1.0 / (1.0 + np.exp(-z))

# 2-4-1 MLP trained from scratch in NumPy
W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

lr = 2.0
for epoch in range(10000):
    a1 = sigmoid(np.dot(X, W1) + b1)
    a2 = sigmoid(np.dot(a1, W2) + b2)
    d2 = (a2 - y) * a2 * (1 - a2)
    d1 = np.dot(d2, W2.T) * a1 * (1 - a1)
    W2 -= lr * np.dot(a1.T, d2)
    b2 -= lr * np.sum(d2, axis=0, keepdims=True)
    W1 -= lr * np.dot(X.T, d1)
    b1 -= lr * np.sum(d1, axis=0, keepdims=True)

print("Predictions after 10000 epochs:")
for i in range(4):
    print(f"{X[i]} -> {a2[i,0]:.4f} (target {y[i,0]})")
```

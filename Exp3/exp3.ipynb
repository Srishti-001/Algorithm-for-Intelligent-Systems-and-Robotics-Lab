```
import collections
import numpy as np

# BFS and DFS differing in exactly a single pop/popleft line!
def search(maze, start, goal, mode='bfs'):
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    frontier = collections.deque([start])
    visited = {start: None}
    expanded = []
    
    while frontier:
        current = frontier.popleft() if mode == 'bfs' else frontier.pop()
        expanded.append(current)
        if current == goal:
            break
        r, c = current
        for dr, dc in moves:
            nr, nc = r + dr, c + dc
            if 0 <= nr < maze.shape[0] and 0 <= nc < maze.shape[1]:
                if maze[nr, nc] == 0 and (nr, nc) not in visited:
                    visited[(nr, nc)] = current
                    frontier.append((nr, nc))
    return expanded, visited

print("Unified BFS/DFS graph search implementation verified.")
```

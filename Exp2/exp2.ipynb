```
import itertools
import time

def is_safe(board, row, col):
    for c in range(col):
        r = board[c]
        if r == row or abs(r - row) == abs(c - col):
            return False
    return True

def solve_queens_bt(n):
    solutions = []
    nodes_visited = 0
    board = [-1] * n

    def backtrack(col):
        nonlocal nodes_visited
        nodes_visited += 1
        if col == n:
            solutions.append(list(board))
            return
        for row in range(n):
            if is_safe(board, row, col):
                board[col] = row
                backtrack(col + 1)
                board[col] = -1

    backtrack(0)
    return solutions, nodes_visited

for n in range(4, 11):
    t0 = time.perf_counter()
    sols, nodes = solve_queens_bt(n)
    dt = time.perf_counter() - t0
    print(f"n={n:2d} | Solutions: {len(sols):6d} | Nodes Visited: {nodes:8d} | Time: {dt:.6f} s")
```
